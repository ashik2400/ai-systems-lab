```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a tutor. Explain at a {level} level."),
    ("human", "Explain {topic} in 3 sentences."),
])

msgs = prompt.invoke({"topic": "embeddings", "level": "beginner"})
print(msgs)
```

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    title: str = Field(description="The movie's title")
    year: int = Field(description="Release year")
    genre: str = Field(description="Main genre")
```

```python
Movie(title="Inception", year=2010, genre="Sci-Fi")   # works
Movie(title="Inception", year="2010", genre="x") # raises ValidationError
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="openai/gpt-oss-120b", temperature=0)
structured_model = model.with_structured_output(Movie)

result = structured_model.invoke("Nolan's Inception came out in 2010, a sci-fi thriller.")
print(result)          # Movie(title='Inception', year=2010, genre='Sci-Fi')
print(result.year + 1) # a real int, so arithmetic works
```

```python
from langchain_core.output_parsers import StrOutputParser

chain = prompt | model | StrOutputParser()

print(chain.invoke({"topic": "embeddings", "level": "beginner"}))
```
